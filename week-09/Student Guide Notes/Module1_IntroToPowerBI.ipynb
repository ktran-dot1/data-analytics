# Module 1 Intro to Power BI

## Section 1.A Getting Started with Power BI

### 1.A.1 What is Power BI?

- Power Bi is a business intelligence solution that enables end-to-end management of reporting
    - In gestion of data from variety of sources
    - Data cleaning
    - Transformation and reshaping of data
    - Data Modeling
    - Data enhancement
    - Analysis and visualization

- Hosts the data in-memory using a powerful database engine called xVelocity
- Saves analysis projects as .pbix files
    - .pbix files needs to be "refreshed" or it will go stale
    - problem when sharing files


### 1.A.4 Power Query and M Language

- Provides tooling to bring data into Power BI project to be modeled
- M language used for data cleaning and transformation

### 1.A.6 xVelocity Analytical Engine

- An **in-memory columnar database engine** that underpins Power BI
    - Stores and hosts the data model
    - DAX is an expression language for xVelocity


### 1.A.7 Columnar Databases

- Columnar Databases are ideal for analytical reporting
- Stores data grouped by the columns it belongs to
    - Useful in scenarios where you plan to read every row to perform an evaluation

### 1.A.8 Data Analysis eXpression Language

- **DAX** stands for Data Analysis eXpressionn Language
- Used to add meaning to the data through formulas, or expressions
- Adds four types of Calculation to our Model
    - Calculated Columns
    - Measures
    - Calculated Tables
    - Row Filters
    

## Section 1.B Power BI Tools

### 1.B.5 The View Selector

- Selection of one of four views
    - **Report View**: used to design and view visual report elements
    - **Table View**: used to view data; some modeling actions can be performed
    - **Model View**: used to configure and see relationship between datasets
    - **DAX Query View**: allows querying of the semantic model using DAX queries
