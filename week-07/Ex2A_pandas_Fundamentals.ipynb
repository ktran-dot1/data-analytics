Create a new variable df1 and give it the value of a DF using ser1

```python
df1.insert(3, "ser4", ser4)
df1

# After calling the DataFrame, the values in column: ser4, all shows NaN. This happened because ser4 has a dictionary collection in it. It has string values that don't match the other values in the DF
```
