# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1,2,3,4,5]
y = [2,3,5,7,11]
```

Use plt.plot() to plot the line graph for x and y, and then display it using plt.show().

```python
plt.plot(x, y)
plt.show()
```

Using the previous code cell, add markers to it

```python
plt.plot(x, y, marker = 'o')
plt.show()
```

Add the title, x-axis label, and y-axis label

```python
plt.plot(x, y, marker='o')
# title
plt.title('Example Graph') 
# x-axis label
plt.xlabel('X axis')
# y-axis label
plt.ylabel('Y axis')

plt.show()
```

## Styling Plots

Change line style to dashed and line color to red

```python
plt.plot(x , y, marker='o', ls='--', c='red')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

Marker style to square, marker size to 10, and marker color to green

```python
plt.plot(x , y, marker='s', ms=10, mfc='g', ls='--', c='red')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

Experiment with at least two different marker style and colors, and display the plot

```python
plt.plot(x , y, marker='*', ms=10, mfc='b', ls='--', c='red')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
plt.plot(x , y, marker='d', ms=10, mfc='purple', ls='--', c='red')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

# Exercise 3.B More plotting practice

## Advance Plotting Techniques

Create a bar chart showing different sales of different products.

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23,17,35,29]

plt.bar(products, sales)
```

Add a title, x-axis label, y-axis label. Set the bar color to 'skyblue' and bar width to 0.5

```python
plt.bar(products, sales, color='skyblue', width=0.5)
plt.title('Products Sold')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

Create a scatter plot to visualize the relationship between two variables.

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
```

```python
s = np.random.rand(50) * 100
plt.scatter(x, y, s=s, c=s, cmap='viridis')
plt.colorbar()

```

## Subplots

Create a figure with two subplots: one for a line chart and one for a histogram

```python
x = np.linspace(0,10,100)
y = np.sin(x)
data = np.random.randn(1000)

# Create the figure with two subplots
fig, (ax1, ax2) = plt.subplots(1,2)

ax1.plot(x, y)
ax2.hist(data, edgecolor='black')

ax1.set_title('Sine Wave')
ax2.set_title('1000 Random Numbers')

ax1.set_xlabel('X axis')
ax2.set_xlabel('X axis')

ax1.set_ylabel('Y axis')
ax2.set_ylabel('Y axis')
plt.tight_layout()
plt.show()
```
