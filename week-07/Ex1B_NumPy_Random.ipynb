## Generating Random Numbers

## Generating Random Choices from an Array

Variable: student, and assign it a 1D array containing the names of 12 of your classmates

```python
students = np.array(['Kevin', 'Brandon', 'Luis', 'Edmund',
                    'Dahlia', 'Destiny', 'Stephanie', 'Natalia',
                    'Jeremiah', 'Hritik', 'John', 'Zack'])
```

Variable: call_on. Use it to choose one random selection from students, Then, print

```python
call_on = np.random.choice(students)
print(f"The person to start the morning standup will be: {call_on}")
```

Variable: breakout_groups. Use it to choose a random sampling from students that is 3 rows, 4 columns

```python
breakout_groups = np.random.choice(students, size = (3, 4))
print(breakout_groups)
# This would not be effective in assigning breakout groups, because there are chances of names repeating, making it not useable
```

Variable: call_on_weighted. Make five random selections from students, but this time assign probability values. One student at least has zero liklihood of being selected, and at least one student with 30% chance of being chosen.

```python
call_on_weighted = np.random.choice(students, size = (5),
                                    p = [.3, .1, .1, .05, .05, .1,
                                         .00, .05, .05, .05, .1, .05])
print(call_on_weighted)

# Kevin has 30% of being selected
# Stephanie has 0% of being selected
# Everyone else has either 10% or 5% of being chosen
```

## Generating a Normal Distribution

Variable: bell_curve. Assign it a random distribution for which the mean is 10, scale = 5, and the size = (4, 5)

```python
bell_curve = np.random.normal(loc = 10, scale = 5, size = (4, 5))
print(bell_curve)
```
