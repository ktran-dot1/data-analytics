## Cumulative Operations

```python
# print r_4
print(r_4)
```

```python
# the sum of all values in r_4
print(np.sum(r_4))
```

```python
# Sum of all elements along the 0 axis in r_4
print(np.sum(r_4, axis = 0))
```

```python
# Now along the 1 axis
print(np.sum(r_4, axis = 1))
```

```python
# cumulative sum of values in r_4
print(np.cumsum(r_4))
```

```python
# Now along the 0 axis
print(np.cumsum(r_4, axis = 0))
```

## Grocery Problem

```python
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
```

```python
total1 = np.sum(bag1)
total2 = np.sum(bag2)

print(f"The first bag contains {total1} items, and the second bag contains {total2} items.")
```

```python
category_total = np.add(bag1, bag2)
print(category_total)
# [Veg, Fruits, Snacks, Bev]

print(f"There are {category_total[0]} vegetables, {category_total[1]} fruits, {category_total[2]} boxes of snackes, and {category_total[3]} beverages in the two grocery bags")
```

## Teamwork

```python
# [Monday, Tuesday, Wednesday, Thursday, Friday]
ali_knocks = np.array([5, 4, 6, 4, 2])
```

```python
ben_knocks = np.array([0, 5, 5, 3, 4])
```

```python
carla_knocks = np.array([3, 6, 0, 3, 4])
```

```python
all_knocks = np.array([ali_knocks, ben_knocks, carla_knocks])
print(all_knocks)
```

```python
personal_total = np.sum(all_knocks, axis = 1)
print(personal_total)

group_total = np.sum(all_knocks)
print(group_total)
```

```python
daily_knocks = np.sum(all_knocks, axis = 0) # axis = 0, so it gets monday - friday
print(daily_knocks)

rolling_knocks = np.cumsum(daily_knocks) * 0.5
print(rolling_knocks) # About 27 votes through the week.
```
