# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

Create a DataFrame named df_sales using the following data:

```python
data = {'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East',
    'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]}

df_sales = pd.DataFrame(data)
df_sales
```

Add a new column Revenue to df_sales by multiplying Units_Sold and Price

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

Use the sum function to calculate the total Units_Sold and total Revenue across all
products and regions

```python
df_sales[['Units_Sold', 'Revenue']].agg('sum')
```

Use the mean function to calculate the average Units_Sold and average Revenue.

```python
df_sales[['Units_Sold', 'Revenue']].agg('mean').round(2)
```

## Grouping Data

Use the groupby function to group df_sales by Product and calculate the total
Units_Sold and total Revenue for each product

```python
df_sales.groupby(['Product'])[['Units_Sold', 'Revenue']].agg('sum')
```

Use the groupby function to group df_sales by Region and calculate the average Units_Sold and average Revenue for each region.

```python
df_sales.groupby(['Region'])[['Units_Sold', 'Revenue']].agg('mean')
```

Use the groupby function to group df_sales by both Product and Region and calculate the sum of Units_Sold and Revenue for each combination

```python
df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].agg('sum')
```

## Sorting and Ranking Data

Sort df_sales by Units_Sold in ascending order

```python
df_sales.sort_values('Units_Sold', ascending = False)
```

Sort df_sales by Revenue in descending order

```python
df_sales.sort_values('Revenue', ascending = False)
```

Add a new column Revenue_Rank to df_sales, which ranks the Revenue values in descending order (i.e., highest revenue has rank 1).

```python
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending = False, method = 'dense').astype(int)
df_sales['Revenue_Rank']
df_sales
```

## Merging DataFrames

Create a new DataFrame named df_products with the following data:

```python
product_data = {'Product': ['A', 'B', 'C'],
'Category': ['Electronics', 'Furniture', 'Appliances']}
df_products = pd.DataFrame(product_data)

df_products = pd.DataFrame(product_data)
df_products
```

Merge df_sales with df_products on the Product column, resulting in a new DataFrame named df_merged

```python
df_merged = pd.merge(df_sales, df_products, on = ['Product'])
df_merged
```

Use groupby on df_merged to calculate the total Revenue by Category

```python
df_merged.groupby(['Category'])['Revenue'].agg('sum')
```
